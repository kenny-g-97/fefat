## Maximum drawdown

The profit and loss of six trades: 100, -40, 30, -120, 50, -10. Their cumulative sum, starting from 0, is 100, 60, 90, -30, 20, 10. Its highest point is 100, and the deepest fall below it is to -30: a maximum drawdown of 130.

Expected output of the next cell:

```text
130
```

````python
pnl = pl.Series("pnl", [100, -40, 30, -120, 50, -10], dtype=pl.Int64)
print(ff.stats.max_drawdown(pnl))
````
