## Maximum drawdown

The profit and loss of six trades: 100, -40, 30, -120, 50, -10. The account's profit after each, their cumulative sum, is 100, 60, 90, -30, 20, 10. Its running drawdown is how far each value lies below the highest one so far, 100: 0, 40, 10, 130, 80, 90. The largest, the maximum drawdown, is 130.

Expected output of the next cell:

```text
[0, 40, 10, 130, 80, 90]
130
```

````python
pnl = pl.Series("pnl", [100, -40, 30, -120, 50, -10], dtype=pl.Int64)
drawdown = ff.backtest.running_drawdown(pnl.cum_sum())
print(drawdown.to_list())
print(drawdown.max())
````
